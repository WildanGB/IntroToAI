## What is NumPy?

Python lists are flexible but slow for numerical work. When you write `[x * 2 for x in data]`, Python loops through each element one at a time. **NumPy** (Numerical Python) stores numbers in a compact block of memory and operates on the entire block at once, the same way your calculator handles a matrix multiplication in one step.

```python
# Python list: loop through one at a time (slow)
result = [x * 2 for x in data]

# NumPy array: operate on the whole block at once (fast)
result = data * 2
```

For 1,000 elements this doesn't matter. For the image arrays and datasets you'll work with in Weeks 8–11, it makes a huge difference.

> **Want to go deeper?** NumPy achieves its speed through *contiguous memory layout* and *SIMD instructions* (the CPU processes multiple numbers in a single clock cycle). If you're curious about the internals, see the [NumPy documentation on array internals](https://numpy.org/doc/stable/dev/internals.html). For FIT1061, you just need to know *how to use* arrays, not how they work under the hood.

## Exercise 2: Indexing and slicing

NumPy uses `[row, col]` indexing. You can slice ranges with `start:stop`.

```python

scores[:, 1]    # Column 1 (Exam 2),       all rows
scores[1, 2]    # Row 1, Column 2,          Bob's Exam 3 score
scores[0:2, :]  # Rows 0-1 (Alice and Bob), all columns
```

````python
# --- YOUR CODE HERE ---

# 2a. Extract Carol's scores (row 2, all columns)
carol_scores = scores[2, :]  # Replace

# 2b. Extract all Exam 1 scores (all rows, column 0)
exam1_scores = scores[:, 0]  # Replace

# 2c. Extract Bob's Exam 3 score (single value)
bob_exam3 = scores[1, 2]  # Replace

# 2d. Extract the top-left 2×2 block (Alice+Bob, Exams 1+2)
top_left = scores[:2, :2]  # Replace

# --- END YOUR CODE ---

print(f"Carol's scores: {carol_scores}")
print(f"Exam 1 scores:  {exam1_scores}")
print(f"Bob's Exam 3:   {bob_exam3}")
print(f"Top-left 2×2:\n{top_left}")
````

## Exercise 3: Aggregation operations

NumPy can compute `sum`, `mean`, `max`, `min` across the entire array or along a specific axis:

- `axis=0`: operate **down** the rows (one result per column)
- `axis=1`: operate **across** the columns (one result per row)
- No axis: operate on **all** elements

```
           Exam1  Exam2  Exam3
  Alice  [  85     90     78  ]  ← axis=1 (mean of this row)
  Bob    [  92     88     95  ]
  Carol  [  76     85     82  ]
  Dan    [  88     79     91  ]
            ↑
         axis=0 (mean of this column)
```

For example ```np.mean(arr, axis=1)``` finds the mean of each row in arr.

````python
# --- YOUR CODE HERE ---

# 3a. Mean score per student (one number per row)
student_means = np.mean(scores, axis=1)  # Replace — use np.mean with axis=

# 3b. Mean score per exam (one number per column)
exam_means = np.mean(scores, axis=0)  # Replace

# 3c. Overall mean (single number)
overall_mean = np.mean(scores)  # Replace

# 3d. Highest score in the entire dataset
highest = np.max(scores)  # Replace

# 3e. Lowest score per exam
exam_mins = np.min(scores, axis=0)  # Replace

# --- END YOUR CODE ---

names = ["Alice", "Bob", "Carol", "Dan"]
print("Mean score per student:")
for name, mean in zip(names, student_means):
    print(f"  {name}: {mean:.1f}")

print(f"\nMean score per exam: {exam_means}")
print(f"Overall mean: {overall_mean:.1f}")
print(f"Highest score: {highest}")
print(f"Lowest per exam: {exam_mins}")
````

## Exercise 4: Boolean indexing

You can use a condition to **filter** array elements. The condition produces a boolean array, which you use as an index.

```python
arr = np.array([10, 25, 30, 15, 40])
mask = arr > 20          # [False, True, True, False, True]
filtered = arr[mask]     # [25, 30, 40]
```

````python
# --- YOUR CODE HERE ---

# 4a. Which students have a mean score above the overall mean?
#     (Use student_means and overall_mean from Exercise 3)
above_average_mask = student_means > overall_mean  # Replace — a boolean array

# 4b. How many students are above average?
num_above = np.sum(above_average_mask)  # Replace — count how many are True in the mask

# 4c. Extract all individual scores that are 90 or above
high_scores = scores[scores >= 90]  # Replace — boolean index on the scores array

# --- END YOUR CODE ---

print(f"Above-average mask: {above_average_mask}")
for name, above in zip(names, above_average_mask):
    print(f"  {name}: {'above' if above else 'below'} average")
print(f"\nNumber above average: {num_above}")
print(f"All scores >= 90: {high_scores}")
````

## Exercise 5: Vectorised operations

NumPy lets you do arithmetic on entire arrays **without loops**. This is called **vectorisation**: it's faster and more readable.

```python
# Without NumPy (slow, verbose):

# With NumPy (fast, one line):
result = data * 2 + 1
```

````python
# --- YOUR CODE HERE ---

# 5a. Normalise each score to the 0-1 range, using the overall
#     minimum and maximum of the scores array.
normalised = (scores - np.min(scores)) / (np.max(scores) - np.min(scores))  # Replace — one line, no loops

# 5b. Curved scores: add 5 points to every score, then cap the result at 100.
curved = scores + 5
curved[curved > 100] = 100  # Replace — one line, no loops
#this wasnt one line but its the only way i could think of without extra functions (although i can make it one line with a semicolon)
curved = np.minimum(scores + 5, 100) #this is one line but i searched up this function, not sure which one was intended

# 5c. Percentage change from Exam 1 to Exam 3 for each student.
#     Extract the Exam 1 and Exam 3 columns, then compute the percentage
#     change from the first to the third.
pct_change = ((scores[:, 2] - scores[:, 0]) / scores[:, 0]) * 100  # Replace

# --- END YOUR CODE ---

print("Normalised scores (0-1 range):")
print(normalised)

print(f"\nCurved scores (+5, max 100):")
print(curved)

print(f"\nPercentage change, Exam 1 → Exam 3:")
for name, pct in zip(names, pct_change):
    print(f"  {name}: {pct:+.1f}%")
````
