## 3. Three Supporting Points

- Claim: Search algorithsm like like GBFS disregard and dont track the accumulated path, resulting in incorrect pathing due to misleading heuristics*
- Evidence: The 15x15 maze grid with corridors and dead-ends from P3: GBFS only expanded about 23 nodes versus the 67 expanded by BFS, but returned a path of 53 steps rather than the optimal 41 step path - a ~30% increase. Likewise, if the size of the obstacle grid is 10x10, Greedy would have found a 23-step path but the optimal one was 19 steps.
- Why it supports your thesis:  It demonstrates that for such a problem, namely the one that we are trying to solve, the quality of the solution is greatly diminished when the focus is on speed instead of g(n).

**Point 2:**
- Claim: The motion engages in a false dichotomy fallacy when implying that fast and optimal are mutually exclusive; in reality the presence of a dominant admissable heuristic guarantees optimality with comparable time as compared to GBFS.
- Evidence: In D1, I used 18 shuffles of 12 random 8-puzzle instances (seed=42) to test BFS, GBFS, and A*. A* Manhattan expanded an average of 10.2 nodes in 0.0002s (beating GFS at 10.4 nodes and 0.0002s) while BFS expanded an average of 250.7 nodes in 0.002s and still got an optimal mean path of 6.5 moves.
- Why it supports your thesis: By using a tighter, consistent heuristic where h2(n) >= h1(n) (Manhattan distance dominating misplaced tiles at 10.2 vs 20.0 nodes), we prune irrelevant branches early and get the exact optimal solution just as fast as a "good-enough" algorithm.

**Point 3:**
- Claim: Whether speed or optimality is preferable is entirely situational on the application and domain; in high-stakes, resource-constrained, or safety-critical systems, sub-optimal solutions cause compounding financial, environmental, or human harm.
- Evidence: Referring to the Week 3 A* supplement, using an inadmissible heuristic (h(n) = 3 * Manhattan)  or greedy search in a route-planning system that promises shortest distance is "malpractice".In airline fuel routing, hospital patient/organ allocation, or Green AI (Schwartz et al.), a 25–30% sub-optimal execution path wastes massive physical resources every single time the plan is executed.
- Why it supports your thesis: The word "always" in the motion fails because whenever the real-world cost of executing extra steps outweighs the one-off CPU cost of finding the plan, guaranteed optimality is strictly superior.

## 5. Your Response to the Counterargument

How do you address this objection without dismissing it? Good responses acknowledge what the other side gets right before explaining why your position still holds.

**I acknowledge that:** This is a perfectly valid hardware constraint. Video games or large systems rely on tight memory management and so when a state exponentially increases in a short number of frames, the search will indoubitably fail due to memroy bottlenecks. In specific cases like AlphaGo using Monte Carlo Search Tree rollouts or real-time videogames, it is necessary to prioritize speed - even at the cause of optimality.

**However:** The exception does not make the rule. The motion was specific on good-enough algorithms being "always" better and these edge cases do not represent the population sample. Furthermore, the aforementioned compromises are forced by hardware restraints and not because suboptimal algorithms are intrinsically more desirable. Even in resource-scarce settings, modern AIs use variations of A* (Anytime A*, etc) that start with a quick path and recursively refine it to the optimal path - this proves that optimality is in fact the main goal when resources allow it.
