## P10.1: Reflective Essay (~500 words)

Write a reflective essay responding to:

> *"Which algorithm was hardest? Which milestone resonated most? What challenged your thinking about AI?"*

Use the prompts below to help structure your thinking. You don't need to answer every question; pick the ones most relevant to your experience.

### What good reflective writing looks like

Here is a vague sentence and a stronger version:

- **Vague:** "BFS was hard."
- **Specific:** "BFS was hard because I kept confusing the queue with a stack; it wasn't until I traced the Kasparov grid by hand in P2 that I realised the frontier expands level by level, not depth-first."

The difference: the specific version names *what* was hard, *why*, and *what changed*. Aim for this level of detail. One well-explained moment is worth more than a list of everything you did.

**One way to structure your essay:** start with the hardest moment, then describe a breakthrough, then look forward. But find the structure that works for you.

### Looking back at the algorithms

- Which algorithm (BFS, DFS, greedy best-first, hill climbing, Naive Bayes, weighted sums, gradient descent) was the hardest for you? What made it difficult: the maths, the code, or the concept?
- Which algorithm "clicked" most naturally? Why?
- Did any hand-trace exercise change how you understood an algorithm?

### Looking back at the milestones

- Which case study (Deep Blue, AlphaGo, COMPAS, the AI Winter, ImageNet) stayed with you most? Why?
- Did any ethical reflection make you reconsider something you previously took for granted?
- Was there a moment where you saw a connection between two different weeks?

### Looking forward

- Look back at your **P1 learning plan**. How did your goals evolve? Did you achieve what you set out to do?
- Has this unit changed how you think about AI in your daily life? Name one AI system you interact with (recommendation algorithm, voice assistant, autocorrect) and explain how your understanding of it changed after a specific week.
- If you could go deeper into one topic, what would it be?
- What advice would you give to a student starting this unit next year?

> **This is not a test.** There are no right or wrong answers. We're looking for honest reflection that shows you've engaged with the material. A thoughtful essay about struggling with gradient descent and eventually understanding it is better than a polished essay that says nothing personal.

**Your essay:**

I came into this unit with the perspective of heuristics and agents in artificial intelligence. Even though my degree is in Data Science, apart from big words like deep learning and unlabeled learning, I had no idea how AIs work and have worked on a fundamental level. In the first few weeks my prior conceptions of weight based learning and graphs were reinforced, but starting Week 10 they were challenged, when we tried gradient descent in P9. Whereas with other search implementations such as BFS and DFS, the states were neatly indexed nodes on a grid, with gradient descent I had to deal with continuous numerical optimization. In my first implementation, the cost function grew very quickly to infinity as I had an unnecessarily high learning rate alpha and implemented it as an arbitrary hyperparameter rather than a scaling factor for the gradient that is sensitive. Only after putting the expression for the mean squared error for a toy model with 2 parameters into code and doing some manual computation of the partial derivatives did I get a sense of what's going on: the gradient is simply a vector that is pointing in the direction of steepest ascent, and subtracting it is taking a step downhill. After tuning alpha, finally the loss curve came to stay flat, an intimidating calculus equation turned to an intuitive geometric view.

The most important challenges for my thinking were the Week 7 milestones on the COMPAS recidivism risk algorithm and the ProPublica investigation. Before I got to that task, I assumed that algorithmic systems are objective; that using a statistical system replaces human judgment and prejudice would go away. The COMPAS case came to eliminate this belief however. The analysis showed that fairness is not easily codifiable into a neat mathematical equation, or that it is impossible to reconcile false positive and false negative error rates between groups having varying base rates. It made me think about algorithms and how they don't exist in an absolute moral vacuum. Finally, machine learning models simply reflect and reinforce the patterns that already exist in their training distribution, and frequently disguise systemic bias as the neutral facade of mathematics.

The aspect that seemed to stick most for the 12 weeks was finding the “conceptual connections” between seemingly unrelated topics. It was such as applying random restart to hill climbing algorithms that it felt like an ad hoc idea for combinatorial puzzles such as the 8-queens puzzle in P4 - I particularly liked this format of using historical examples to showcase how AI technology developed and re-implementing those technologies to reveal the underlying methodology. However, during Week 10 in machine learning, the same theme came up with the exploring of loss surfaces which are not convex. Both discrete search and continuous learning are essentially the same thing: both are searching through very large parameter spaces, based on an evaluation metric.

When I look at my learning plan for Week 1, I have changed my focus from making models that work to challenging the models underpinning them and the societal costs of the models - while I knew of the environmental concerns, C4 really put it into perspective for me. I don't feel the same way about the use of common predictive tools and recommendation engines anymore than when I use them. To someone who is just starting out this year I do not recommend immediately launching into coding. Mark the boundaries on a piece of paper, derive rhe error gradients by hand and never take anything at face value from the objective function. AI is a buzzword in today's world, but it is worth understanding the why to gain a grasp of how far this subject has come and what it yet has to deliver.
